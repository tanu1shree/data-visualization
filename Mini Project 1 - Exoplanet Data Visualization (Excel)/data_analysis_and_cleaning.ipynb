# Data Cleaning

```python
df.isna().sum()
```

```python
# replacing any invalid infinite values with NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
df.duplicated().sum()
```

```python
Q1 = df["radius"].quantile(0.25)
Q3 = df["radius"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["radius"] < lower) | (df["radius"] > upper)]

len(outliers)
```
